```python
from mbdd.circuit import *
from mbdd.parser import *
from mbdd.comparator import *
```

```python
NUM_QUBITS = 4
# NUM_OPERATIONS = 20
INIT_MATRIX_SIZE = 2 * NUM_QUBITS + 1

# operations = randomizer(option=ControlledGate)

operations : List[GateOperation] = []
operations.append(operation_constructor(GateType.H, 3))
operations.append(operation_constructor(GateType.H, 2))
operations.append(operation_constructor(GateType.H, 0))
operations.append(operation_constructor(GateType.H, 1))
operations.append(operation_constructor(GateType.X, 1, [0, 2, 3]))
operations.append(operation_constructor(GateType.X, 1, [0]))
operations.append(operation_constructor(GateType.X, 1, [2, 3]))
operations.append(operation_constructor(GateType.H, 1, [0, 2, 3]))

operations.append(operation_constructor(GateType.X, 3, [1, 2]))

operations.append(operation_constructor(GateType.H, 0, [1, 2]))
operations.append(operation_constructor(GateType.H, 0, [1, 2]))
operations.append(operation_constructor(GateType.H, 0, [1, 2]))




# _view_operations(operations)
```

```python
strong_result = dd_sparse_numpy(operations, INIT_MATRIX_SIZE, NUM_QUBITS, MODE='strong_simulation')
```

```python
weak_result = dd_sparse_numpy(operations, INIT_MATRIX_SIZE, NUM_QUBITS, MODE='weak_simulation')
```

```python
print("Strong Simulation Result:")
display(strong_result)
print("Weak Simulation Result:")
display(weak_result)
```

```python
compare_strong_weak(strong_result['sim_result'], weak_result['sim_result'])
```
